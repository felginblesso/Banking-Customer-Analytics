# EDA

```python
# Feature Engineering (Categorized customer based on their estimated Income)

bins = [0, 100000, 500000, float('inf')]
labels = ['Low', 'Medium', 'High']

df['Income Band'] = pd.cut(df['Estimated Income'], bins= bins, labels= labels, right= False)
```

```python
df['Income Band'].value_counts().reindex(['Low', 'Medium', 'High']).plot(kind= 'bar')

plt.ylabel('Count')
plt.xlabel('Income Band')
plt.title('Customer Count By Income Band')
plt.show()
```

```python
# Distribution of customer across nations

df['Nationality'].value_counts().plot(kind= 'bar')

plt.ylabel('Count')
plt.xlabel('Nation')
plt.title('Customer Distribution By Nation')
plt.show()
```

```python
# Credit card usage

df['Amount of Credit Cards'].value_counts().plot(kind= 'bar')

plt.ylabel('Count')
plt.xlabel('Credit Card')
plt.title('Credit Card Used BY Customers')
plt.show()
```

```python
# Gender Distribution

gender_labels = {1: "Male", 2: "Female"}

df['GenderId'].map(gender_labels).value_counts().plot(kind= 'bar')

plt.ylabel('Count')
plt.xlabel('Gender')
plt.title('Gender Distribution')
plt.show()
```

```python
# Banking Relationship

banking_relationship = {1: "Retail", 2: "Institutional", 3: "Private Bank", 4: "Commercial"}

df['BRId'].map(banking_relationship).value_counts().plot(kind= 'bar')

plt.ylabel('Count')
plt.xlabel('Banks')
plt.title('Banking Relationship')
plt.show()
```

```python
sns.countplot(
    x='Risk Weighting',
    data=df,
    order=sorted(df['Risk Weighting'].unique())
)

plt.title('Distribution of Risk Weighting')
plt.xlabel('Risk Weighting')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='Estimated Income', y='Bank Loans')

plt.xlabel('Estimated Income')
plt.ylabel('Bank Loans')
plt.title('Estimated Income vs Bank Loans')
plt.show()

"""
Notes:
    - Most of the low income people takes small loan amount
    - Only few peoplw with low income takes a higher loan amount
    - Higher income people takes a higher loan amount
"""
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data= df, x= 'Income Band', y= 'Bank Loans')

plt.xlabel("Income Band")
plt.ylabel("Bank Loan")
plt.title("Bank Loan by Income Band")
plt.show()
```

```python
loan_customers = df[df['Bank Loans'] > 0]

sns.countplot(
    data=loan_customers,
    x='Income Band'
)

plt.xlabel('Income Band')
plt.ylabel('Number of Customers')
plt.title('Loan Customers by Income Band')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Amount of Credit Cards',
    y='Bank Loans'
)

plt.xlabel('Number of Credit Cards')
plt.ylabel('Bank Loans')
plt.title('Bank Loans by Number of Credit Cards')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Income Band',
    y='Risk Weighting'
)

plt.xlabel('Income Band')
plt.ylabel('Risk Weighting')
plt.title('Risk Weighting by Income Band')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Loyalty Classification',
    y='Bank Loans'
)

plt.xlabel('Loyalty Classification')
plt.ylabel('Bank Loans')
plt.title('Bank Loans by Loyalty Classification')
plt.show()
```

```python

columns = [
    'Age',
    'Estimated Income',
    'Superannuation Savings',
    'Credit Card Balance',
    'Bank Loans',
    'Bank Deposits'
]

# Draw histogram for each column
for column in columns:
    plt.figure(figsize=(6, 4))
    plt.hist(df[column], bins=30, edgecolor='black')
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.show()
```

```python
columns = [
    'Estimated Income',
    'Superannuation Savings',
    'Amount of Credit Cards',
    'Credit Card Balance',
    'Bank Loans',
    'Bank Deposits',
    'Checking Accounts',
    'Saving Accounts',
    'Foreign Currency Account',
    'Business Lending',
    'Properties Owned',
    'Risk Weighting'
]

# Create correlation matrix
corr = df[columns].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Heatmap of Banking Features')
plt.show()
```
